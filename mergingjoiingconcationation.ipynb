```python
import numpy as np
import pandas as pd
```

```python
employees = pd.DataFrame({
    'employee_id': [1,2,3,4,5],
    "Name": ['rishi','hardyansh','mohit','kinjal',"Lokesh"],
    "daprtment": ["IT",'HR','FINANCE','IT','HR'],
})


salaries = pd.DataFrame({
    'employee_id': [1, 2, 3, 6, 7],
    'salary': [60000, 80000, 65000, 70000, 90000],
    'bonus': [5000, 10000, 7000, 8000, 12000]
})
```

```python
employees
```

```python
salaries
```

```python
merge 2 data frames
```

```python
salaries
```

```python
pd.merge(employees,salaries,on="employee_id",how="inner")
```

```python
pd.merge(employees,salaries, on="employee_id",how="outer")
```

```python
Concatination of Two Frames
```

```python
df1 = pd.DataFrame(
    {
        'A':['A0','A1','A2'],
        'B':['B0','B1','B2'],
        'C':['c1','c2','c3']
    }
)
df2 = pd.DataFrame({
    'A':["A4","A5","A6"],
    'B':["B4","B5","B6"],
    'C':["C4","C5","C6"]
})
```

```python
df1
```

```python
df2
```

```python
pd.concat([df1,df2])
```

```python
pd.concat([df1,df2], 
          axis= 1)
```

```python
JOINING 2 DATAFRAMES
```

```python
df1 = pd.DataFrame({
    'Name':['Alice','bob','Charlie']
},index=[1,2,3])

df2 = pd.DataFrame({
    'Score':[89,76,88]
},index=[2,3,4])
```

```python
df1
```

```python
df2
```

```python
df1.join(df2,how="outer")
```

```python
df2.join(df1)
```
