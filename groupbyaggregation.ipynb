```python
import numpy as np
import pandas as pd
```

```python
data = {
    'Category': ['A', 'B', 'A', 'B', 'A', 'B', 'A', 'B'],
    'Store': ['S1', 'S1', 'S2', 'S2', 'S1', 'S2', 'S2', 'S1'],
    'Sales': [100, 200, 150, 250, 120, 180, 200, 300],
    'Quantity': [10, 15, 12, 18, 8, 20, 15, 25],
    'Date': pd.date_range('2023-01-01', periods=8)
}
df = pd.DataFrame(data)
```

```python
cat = df.groupby('Category')
for i, b in cat:
    print(i)
    print(b)
```

```python
#group by category and calculate the sum of the sales
cate = df.groupby('Category') ['Sales'].sum()
cate
```

```python
#Group by store and calulate the sum of the sales
st = df.groupby('Store') ['Sales'].sum()
st
```

```python
#group by multiple columns
#group by category and store
cater = df.groupby(['Category','Store'])['Sales'].sum()
cater
```
